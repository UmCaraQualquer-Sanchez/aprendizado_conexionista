```python

import numpy as np
from numpy.random import randn

# Dados e tamanho da rede
N, D_in, H, D_out = 64, 1000, 100, 10

# Criando os dados e os pesos
x, y = randn(N, D_in), randn(N, D_out)
w1, w2 = randn(D_in, H), randn(H, D_out)

# Treinamento da rede
for t in range(2000):
    # Calculando a saída da rede
    h = 1 / (1 + np.exp(-x.dot(w1)))
    y_pred = h.dot(w2)

    # Calculando o erro
    loss = np.square(y_pred - y).sum()

    if t % 100 == 0:
        print(t, loss)

    # Calculando os gradientes
    grad_y_pred = 2.0 * (y_pred - y)
    grad_w2 = h.T.dot(grad_y_pred)
    grad_h = grad_y_pred.dot(w2.T)
    grad_w1 = x.T.dot(grad_h * h * (1 - h))

    # Atualizando os pesos
    w1 -= 1e-4 * grad_w1
    w2 -= 1e-4 * grad_w2

# Mostrando as camadas da rede
print("Camada de entrada:", D_in)
print("Camada oculta:", H)
print("Camada de saída:", D_out)
```
